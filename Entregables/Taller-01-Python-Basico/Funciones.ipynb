## ¿Qué es una Función?

Una función es un bloque de código reutilizable que realiza una tarea específica.
Se define con la palabra clave def, seguida del nombre y paréntesis con los parámetros.
Puede recibir datos de entrada (parámetros) y retornar un resultado con `return`.

```python
# Función básica sin parámetros
def saludar():
    print("Hola desde una función!")

# Función con parámetros y retorno
def sumar(a, b):
    return a + b

saludar()
print(sumar(5, 3))
```

## Parámetros y Retorno

Las funciones pueden recibir múltiples parámetros y retornar cualquier tipo de dato con `*args` se pueden recibir un número variable de argumentos.

<img src="./imagenes/funciones_parametros.png" width="500"/>

```python
# Múltiples parámetros
def presentar(nombre, lenguaje):
    return f"{nombre} programa en {lenguaje}"

print(presentar("Cesar", "Python"))

# Retornar múltiples valores
def operaciones(a, b):
    return a + b, a - b, a * b

suma, resta, mult = operaciones(10, 3)
print(suma, resta, mult)

# *args - número variable de argumentos
def sumar_todos(*args):
    return sum(args)

print(sumar_todos(1, 2, 3, 4, 5))
```

## Funciones Predefinidas

Python tiene funciones integradas que no necesitan importarse y están disponibles
en cualquier momento. Algunas de las más usadas son:

- `len()`: retorna la longitud de una colección.
- `type()`: retorna el tipo de dato de un objeto.
- `range()`: genera una secuencia de números.
- `map()`: aplica una función a cada elemento de una colección.
- `filter()`: filtra elementos de una colección según una condición.

```python
lenguajes = ["Python", "Java", "Kotlin", "JavaScript", "Swift"]

# len() y type()
print(len(lenguajes))
print(type(lenguajes))

# range()
print(list(range(1, 6)))

# map() - convierte cada elemento a mayúsculas
mayusculas = list(map(str.upper, lenguajes))
print(mayusculas)

# filter() - filtra los lenguajes con más de 4 caracteres
largos = list(filter(lambda x: len(x) > 4, lenguajes))
print(largos)
```

## if/else y Ciclos dentro de Funciones

Las funciones pueden contener cualquier estructura de control como condiciones
y bucles, lo que las hace mucho más poderosas y reutilizables.

<img src="./imagenes/funciones_if_ciclos.png" width="500"/>

```python
# if/else dentro de una función
def clasificar_lenguaje(lenguaje):
    if lenguaje == "Python":
        return "Lenguaje principal del curso"
    elif lenguaje in ["Java", "Kotlin"]:
        return "Lenguaje orientado a objetos"
    else:
        return "Otro lenguaje"

print(clasificar_lenguaje("Python"))
print(clasificar_lenguaje("Java"))
print(clasificar_lenguaje("Rust"))

# Ciclo dentro de una función
def listar_lenguajes(lenguajes):
    for i, lenguaje in enumerate(lenguajes, 1):
        print(f"{i}. {lenguaje}")

listar_lenguajes(["Python", "Java", "Kotlin", "Swift"])
```

## Argumentos con Valores por Defecto

Se pueden definir valores predeterminados para los parámetros de una función, si no se pasa ese argumento al llamar la función, se usa el valor por defecto.

```python
# Valor por defecto en parámetros
def saludar(nombre, lenguaje="Python"):
    return f"Hola {nombre}, tu lenguaje favorito es {lenguaje}"

print(saludar("Cesar"))               # Usa el valor por defecto
print(saludar("Cesar", "Kotlin"))     # Sobreescribe el valor por defecto

# Múltiples valores por defecto
def crear_perfil(nombre, rol="Estudiante", universidad="Javeriana"):
    return f"{nombre} | {rol} | {universidad}"

print(crear_perfil("Cesar"))
print(crear_perfil("Cesar", "Desarrollador", "UNAL"))
```

## Variables Locales y Globales

El ámbito de una variable define dónde puede ser accedida dentro del programa.

- Local: se define dentro de una función y solo existe dentro de ella.
- Global: se define fuera de las funciones y es accesible desde cualquier parte.
- global: palabra clave para modificar una variable global desde dentro de una función.

```python
# Variable global
lenguaje = "Python"

def mostrar_lenguaje():
    # Variable local
    version = "3.14"
    print(f"{lenguaje} {version}")

mostrar_lenguaje()
# print(version)  # Error: version no existe fuera de la función

# Modificar variable global con global
contador = 0

def incrementar():
    global contador
    contador += 1

incrementar()
incrementar()
incrementar()
print(contador)  # 3
```

## Funciones Lambda

Una función lambda es una función anónima de una sola línea, se define con la
palabra clave (lambda) y es útil para operaciones simples sin necesidad de crear
una función completa con def.

- Son funciones de una sola expresión.
- Se usan frecuentemente con `map()`, `filter()` y `sorted()`.

<img src="./imagenes/funciones_lambda.png" width="500"/>


```python
# Lambda básica
cuadrado = lambda x: x ** 2
print(cuadrado(5))

# Lambda con múltiples parámetros
sumar = lambda a, b: a + b
print(sumar(3, 7))

# Lambda con sorted()
lenguajes = ["Python", "Java", "Kotlin", "JavaScript", "Swift"]
ordenados = sorted(lenguajes, key=lambda x: len(x))
print(ordenados)  # Ordenados por longitud

# Lambda con map()
numeros = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(cuadrados)

# Lambda con filter()
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(pares)
```
