## ¿Qué es un cadena?

Una **cadena** o String es una cadena de caracteres encerrada entre comillas simples por lo cual son inmutables, es decir, una vez creada no son modificables directamente.  

```python
# Ejemplos básicos de cadenas
cadena1 = "Hola Mundo"
cadena2 = 'Python es genial'
cadena3 = """Esta es una
cadena multilínea"""

print(cadena1)
print(cadena2)
print(cadena3)
```

## Indexación y Slicing

Las cadenas se pueden acceder por índice. El primer carácter tiene índice `0`, también se puede extraer una porción con la sintaxis `cadena[inicio:fin]`.

![descripcion](./imagenes/cadenas_indexacion.png)

```python
cadena = "Procesamiento"

# Indexación
print(cadena[0])    # Primera letra
print(cadena[-1])   # Última letra

# Slicing
print(cadena[0:5])  # Del índice 0 al 4
print(cadena[5:])   # Desde el índice 5 hasta el final
print(cadena[:5])   # Desde el inicio hasta el índice 4
```

## Indexación Negativa

Con índices negativos se accede a la cadena desde el final hacia el inicio el índice `-1` es el último carácter, `-2` el penúltimo, y así sucesivamente.

![descripcion](./imagenes/cadenas_indexacion_negativa.png) 

```python
cadena = "Procesamiento"

print(cadena[-1])   # Último carácter
print(cadena[-2])   # Penúltimo carácter
print(cadena[-5:])  # Últimos 5 caracteres
print(cadena[:-5])  # Todo menos los últimos 5 caracteres
```

## Métodos de Cadenas

Python tiene métodos integrados para manipular cadenas el cual se llaman con la sintaxis `cadena.metodo()`.

![descripcion](./imagenes/cadenas_metodos.png)

```python
cadena = "hola mundo desde python"

# Mayúsculas y minúsculas
print(cadena.upper())       # Todo en mayúsculas
print(cadena.lower())       # Todo en minúsculas
print(cadena.capitalize())  # Primera letra en mayúscula
print(cadena.title())       # Primera letra de cada palabra en mayúscula

# Búsqueda y reemplazo
print(cadena.find("mundo"))      # Retorna el índice donde encuentra la palabra
print(cadena.replace("hola", "adiós"))  # Reemplaza una palabra por otra

# Limpieza
cadena_espacios = "   hola mundo   "
print(cadena_espacios.strip())   # Elimina espacios al inicio y al final

# Separar y unir
print(cadena.split(" "))         # Separa por espacios, retorna una lista
print("-".join(["hola", "mundo", "python"]))  # Une con guión
```
