## ¿Qué es un Diccionario?

Un diccionario es una colección de pares llave-valor. Funciona como una tabla
donde cada llave es única y apunta a un valor específico. Se define con llaves `{}`
y cada par se separa con dos puntos `:`.

- Las llaves pueden ser cadenas, números o tuplas.
- Los valores pueden ser de cualquier tipo.

<img src="./imagenes/diccionarios_definicion.png" width="500"/>

```python
# Ejemplo básico de diccionario
disco = {
    "artist": "Michael Jackson",
    "album": "Thriller",
    "released_year": 1982,
    "genre": "Pop",
    "sales_millions": 66
}

print(disco)
print(type(disco))

# Acceder a un valor por su llave
print(disco["artist"])
print(disco["released_year"])
```

## Operaciones con Diccionarios

- `keys()`: Retorna todas las llaves del diccionario.
- `values()`: Retorna todos los valores.
- `items()`: Retorna los pares llave-valor.
- `get()`: Obtiene un valor por su llave, sin lanzar error si no existe.
- `update()`: Agrega o modifica un par llave-valor.
- `pop()`: Elimina un par por su llave.

<img src="./imagenes/diccionarios_operaciones.png" width="500"/>

```python
disco = {
    "artista": "Michael Jackson",
    "album": "Thriller",
    "Año de salida": 1982,
    "genero": "Pop"
}

# Llaves, valores y pares
print(disco.keys())
print(disco.values())
print(disco.items())

# get() - no lanza error si la llave no existe
print(disco.get("artista"))
print(disco.get("precio", "No existe"))

# update() - agregar o modificar
disco.update({"sales_millions": 66})
print(disco)

# pop() - eliminar
disco.pop("genero")
print(disco)
```
