## ¿Qué es una Clase?

Una clase es una plantilla para crear objetos estas definen los atributos(datos)
y métodos(funciones) que tendrán los objetos creados a partir de ella estos se define con la palabra clave `class`.

<img src="./imagenes/clases_definicion.png" width="500"/>

```python
# Definir una clase básica
class Lenguaje:
    def __init__(self, nombre, creador, año):
        self.nombre = nombre
        self.creador = creador
        self.año = año

    def describir(self):
        return f"{self.nombre} fue creado por {self.creador} en {self.año}"

# Crear objetos
python = Lenguaje("Python", "Guido van Rossum", 1991)
java = Lenguaje("Java", "James Gosling", 1995)

print(python.describir())
print(java.describir())
```

## Herencia

La herencia permite crear una clase nueva basada en una clase existente la clase nueva hereda los atributos y métodos de la clase padre y puede agregar los suyos propios.

<img src="./imagenes/clases_herencia.png" width="500"/>

```python
# Clase padre
class Lenguaje:
    def __init__(self, nombre, año):
        self.nombre = nombre
        self.año = año

    def describir(self):
        return f"{self.nombre} fue creado en {self.año}"

# Clase hija
class LenguajeWeb(Lenguaje):
    def __init__(self, nombre, año, framework):
        super().__init__(nombre, año)
        self.framework = framework

    def describir(self):
        return f"{self.nombre} usa el framework {self.framework}"

# Crear objetos
javascript = LenguajeWeb("JavaScript", 1995, "React")
python_web = LenguajeWeb("Python", 1991, "Django")

print(javascript.describir())
print(python_web.describir())
```

## Encapsulamiento

El encapsulamiento permite proteger los atributos de una clase para que no sean
accedidos o modificados directamente desde fuera, se usan métodos `get` y `set` para
controlar el acceso a los atributos privados, que se definen con `__` al inicio.

<img src="./imagenes/clases_encapsulamiento.png" width="500"/>

```python
class CuentaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.__saldo = saldo  # Atributo privado

    def get_saldo(self):
        return self.__saldo

    def depositar(self, monto):
        self.__saldo += monto
        return f"Depósito exitoso. Saldo actual: {self.__saldo}"

    def retirar(self, monto):
        if monto > self.__saldo:
            return "Saldo insuficiente"
        self.__saldo -= monto
        return f"Retiro exitoso. Saldo actual: {self.__saldo}"

cuenta = CuentaBancaria("Cesar", 1000)
print(cuenta.get_saldo())
print(cuenta.depositar(500))
print(cuenta.retirar(200))
print(cuenta.retirar(2000))
```
