# Listas en Python

Una lista es una colección ordenada y mutable de elementos, lo que significa que se pueden modificar, agregar o eliminar elementos después de su creación. Se definen con corchetes `[]`.

```python
# Ejemplos básicos de listas
lista1 = [1, 2, 3, 4, 5]
lista2 = ["Python", "Java", "Kotlin"]
lista3 = [True, 3.14, "texto", 42]  # Lista con tipos mixtos

print(lista1)
print(lista2)
print(lista3)
print(type(lista1))
```

## Indexación

Las listas se indexan igual que las tuplas. El primer elemento tiene índice `0`, y también se puede usar índices negativos para acceder desde el final.

```python
lenguajes = ["Python", "Java", "Kotlin", "JavaScript", "Swift"]

print(lenguajes[0])    # Primer elemento
print(lenguajes[-1])   # Último elemento
print(lenguajes[2])    # Tercer elemento
```

## Contenido de la Lista

Las listas pueden contener cualquier tipo de dato: números, cadenas, booleanos, e incluso otras listas. Se puede conocer el número de elementos con `len()` y verificar si un elemento existe con `in`.

```python
lenguajes = ["Python", "Java", "Kotlin", "JavaScript", "Swift"]

# Longitud de la lista
print(len(lenguajes))

# Verificar si un elemento existe
print("Python" in lenguajes)
print("Rust" in lenguajes)

# Listas anidadas
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matriz[0])        # Primera fila
print(matriz[1][2])     # Fila 1, columna 2
```

## Operaciones con Listas

Las listas permiten agregar, modificar y eliminar elementos gracias a su mutabilidad.

```python
lenguajes = ["Python", "Java", "Kotlin"]

# Agregar elementos
lenguajes.append("Swift")
lenguajes.insert(1, "Go")
print(lenguajes)

# Modificar elementos
lenguajes[0] = "Rust"
print(lenguajes)

# Eliminar elementos
lenguajes.remove("Go")
lenguajes.pop()
print(lenguajes)

# Concatenar listas
otros = ["JavaScript", "TypeScript"]
todos = lenguajes + otros
print(todos)
```

## Copiado y Clonación de Listas

Asignar una lista a otra variable no la copia, solo crea una referencia al mismo objeto. Para copiar correctamente se usa `.copy()` o `list()`.


```python
original = ["Python", "Java", "Kotlin"]

# Referencia (NO es una copia)
referencia = original
referencia.append("Swift")
print(original)    # También se modifica

# Copia correcta con .copy()
copia = original.copy()
copia.append("Go")
print(original)    # No se modifica
print(copia)

# Copia con list()
copia2 = list(original)
copia2.append("Rust")
print(original)    # No se modifica
print(copia2)
```
