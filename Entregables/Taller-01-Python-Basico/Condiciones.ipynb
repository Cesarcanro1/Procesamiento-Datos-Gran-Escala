## ¿Qué son las Condiciones?

Las condiciones permiten ejecutar bloques de código según si una expresión es `True` o `False`,
Python usa if, elif y else para controlar el flujo del programa.

```python
edad = 20

if edad >= 18:
    print("Es mayor de edad")
elif edad >= 13:
    print("Es adolescente")
else:
    print("Es menor de edad")
```

## Operadores de Comparación

Los operadores de comparación comparan dos valores y retornan `True` o `False`.



<img src="./imagenes/condiciones_comparacion.png" width="500"/>

```python
a = 10
b = 20

print(a == b)   # False
print(a != b)   # True
print(a > b)    # False
print(a < b)    # True
print(a >= 10)  # True
print(a <= 5)   # False
```

## Operadores Lógicos

Los operadores lógicos permiten combinar múltiples condiciones.


<img src="./imagenes/condiciones_logica.png" width="500"/>

```python
edad = 20
tiene_carnet = True

# and - ambas deben ser True
print(edad >= 18 and tiene_carnet)   # True

# or - al menos una debe ser True
print(edad >= 18 or tiene_carnet)    # True

# not - invierte el resultado
print(not tiene_carnet)              # False
print(not edad < 18)                 # True
```

## Condiciones Anidadas

Las condiciones se pueden anidar, es decir, poner un `if` dentro de otro `if`
para evaluar múltiples niveles de condiciones.

```python
edad = 20
nacionalidad = "colombiana"

if edad >= 18:
    if nacionalidad == "colombiana":
        print("Puede votar en Colombia")
    else:
        print("Es mayor de edad pero no puede votar en Colombia")
else:
    print("No puede votar, es menor de edad")
```
