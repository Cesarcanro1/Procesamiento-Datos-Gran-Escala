## ¿Qué son los Bucles?

Los bucles permiten repetir un bloque de código múltiples veces,
Python tiene dos tipos de bucles: `for` que itera sobre una secuencia
y `while` que repite mientras una condición sea `True`.

```python
# Bucle for itera sobre una lista
lenguajes = ["Python", "Java", "Kotlin", "Swift"]

for lenguaje in lenguajes:
    print(lenguaje)
```

## Bucle for con range()

`range()` genera una secuencia de números. Es muy útil cuando necesitas
repetir algo un número específico de veces.

- `range(n)` genera números del 0 al n-1
- `range(inicio, fin)` genera números del inicio al fin-1
- `range(inicio, fin, paso)` genera números saltando de en paso en paso

```python
# range() básico
for i in range(5):
    print(i)

print("---")

# range con inicio y fin
for i in range(2, 8):
    print(i)

print("---")

# range con paso
for i in range(0, 20, 5):
    print(i)
```

## Bucle while

El bucle `while` repite un bloque de código mientras una condición sea True,
es importante asegurarse de que la condición eventualmente sea False
para evitar un bucle infinito.

```python
# while básico
contador = 0

while contador < 5:
    print(f"Contador: {contador}")
    contador += 1

print("---")

# while con break
numero = 0

while True:
    print(f"Numero: {numero}")
    numero += 1
    if numero == 5:
        break
```

## Break y Continue

Break detiene el bucle completamente cuando se cumple una condición y continue salta la iteración actual y continúa con la siguiente.

```python
# break - detiene el bucle
for i in range(10):
    if i == 5:
        break
    print(i)

print("---")

# continue - salta la iteración
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)  # Solo imprime números impares
```
