## ¿Qué es un Conjunto?

Un conjunto es una colección de elementos únicos y desordenados.
Se define con llaves `{}` y Python elimina automáticamente los elementos duplicados.
A diferencia de las listas y tuplas, los conjuntos no tienen índices.

<img src="./imagenes/conjuntos_definicion.png" width="500"/>

```python
# Ejemplos básicos de conjuntos
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
set2 = {1, 2, 3, 4, 5, 5, 5}  # Los duplicados se eliminan automáticamente
set3 = set()  # Conjunto vacío

print(set1)  # Los duplicados desaparecen
print(set2)
print(type(set1))
print(len(set1))
```

## Operaciones con Conjuntos

Los conjuntos tienen métodos para manipular sus elementos:

- `add()`: agrega un elemento al conjunto, si ya existe no lo duplica.
- `remove()`: elimina un elemento, si no existe lanza un error.
- `discard()`: elimina un elemento, si no existe, no lanza error.
- `in`: verifica si un elemento pertenece al conjunto, retorna True (verdadero) o False (falso).

<img src="./imagenes/conjuntos_operaciones.png" width="500"/>

```python
conjunto = {"Python", "Java", "Kotlin"}

# Agregar elementos
conjunto.add("Swift")
print(conjunto)

# Eliminar elementos
conjunto.remove("Java")   
print(conjunto)

conjunto.discard("Go")    
print(conjunto)

# Verificar si un elemento existe
print("Python" in conjunto)
print("Java" in conjunto)

# Recorrer un conjunto
for lenguaje in conjunto:
    print(lenguaje)
```

## Operaciones Lógicas con Conjuntos

Los conjuntos permiten realizar operaciones matemáticas entre ellos,
similares a las operaciones de la teoría de conjuntos.

- `union()`: Retorna todos los elementos de ambos conjuntos sin duplicados.
- `intersection()`: Retorna solo los elementos que están en ambos conjuntos.
- `difference()`: Retorna los elementos que están en el primero pero no en el segundo.
- `symmetric_difference()`: Retorna los elementos que no se repiten en ninguno de los dos.

<img src="./imagenes/conjuntos_logica.png" width="500"/>

```python
set_a = {"Python", "Java", "Kotlin", "Swift"}
set_b = {"Java", "Swift", "Go", "Rust"}

# Unión
print(set_a.union(set_b))

# Intersección
print(set_a.intersection(set_b))

# Diferencia
print(set_a.difference(set_b))

# Diferencia simétrica
print(set_a.symmetric_difference(set_b))
```
