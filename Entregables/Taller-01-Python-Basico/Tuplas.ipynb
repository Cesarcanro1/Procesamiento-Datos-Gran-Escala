## ¿Qué es una Tupla?

Una tupla es una colección ordenada e inmutable de elementos se define con paréntesis `()` y los elementos se separan por comas a diferencia de las listas, una vez creada no se puede modificar.

```python
# Ejemplos básicos de tuplas
tupla1 = (1, 2, 3, 4, 5)
tupla2 = ("Python", "Java", "Kotlin")
tupla3 = (1, "hola", 3.14, True)  # Puede tener diferentes tipos
tupla4 = (42,)  # Tupla de un solo elemento, la coma es obligatoria

print(tupla1)
print(tupla2)
print(tupla3)
print(tupla4)
print(type(tupla1))
```

## Indexación

Las tuplas se pueden acceder por índice igual que las cadenas y el primer elemento tiene índice `0` y por lo que también se puede usar indexación negativa.

```python
tupla = ("Python", "Java", "Kotlin", "JavaScript", "Swift")

# Indexación positiva
print(tupla[0])   # Primer elemento
print(tupla[2])   # Tercer elemento

# Indexación negativa
print(tupla[-1])  # Último elemento
print(tupla[-2])  # Penúltimo elemento
```

## Slicing

El slicing permite extraer una porción de la tupla usando la sintaxis `tupla[inicio:fin]`, el elemento del índice `fin` no se incluye en el resultado.

```python
tupla = ("Python", "Java", "Kotlin", "JavaScript", "Swift", "Go", "Rust")

print(tupla[0:3])   # Del índice 0 al 2
print(tupla[2:])    # Desde el índice 2 hasta el final
print(tupla[:4])    # Desde el inicio hasta el índice 3
print(tupla[-3:])   # Últimos 3 elementos
print(tupla[::2])   # Cada 2 elementos
print(tupla[::-1])  # Tupla al revés
```

## Ordenamiento

Las tuplas son inmutables, por lo que no se pueden ordenar directamente como las listas. Sin embargo, se puede usar la función `sorted()` que retorna una ista ordenada con los elementos de la tupla, o convertirla a lista, ordenarla y volver a convertirla a tupla.

```python
tupla = (5, 2, 8, 1, 9, 3, 7, 4, 6)

# sorted() retorna una lista ordenada (ascendente por defecto)
ordenada_asc = sorted(tupla)
print(ordenada_asc)

# Orden descendente
ordenada_desc = sorted(tupla, reverse=True)
print(ordenada_desc)

# Convertir a tupla nuevamente
tupla_ordenada = tuple(sorted(tupla))
print(tupla_ordenada)
print(type(tupla_ordenada))

# Ordenar cadenas
lenguajes = ("Python", "Java", "Kotlin", "JavaScript", "Swift")
print(sorted(lenguajes))           # Orden alfabético
print(sorted(lenguajes, reverse=True))  # Orden alfabético inverso
```
